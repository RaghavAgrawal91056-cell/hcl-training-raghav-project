```python
import pandas as pd
df=pd.read_csv('/content/zoo.csv')

#Read
df.head()
```

```python
# 1  How many animals are present in each animal type
print("Total sum of each type of animal :",df['type'].sum())
print("Total number of each type of animal :")
df['type'].value_counts()
```

```python
# 2  How many animals have hair, feathers, eggs or milk
print("Numbers of animal with hair :",df['hair'].sum())
print("Numbers of animal with feathers :",df['feathers'].sum())
print("Numbers of animal with eggs :",df['eggs'].sum())
print("Numbers of animal with milk :",df['milk'].sum())

total=(df[['hair','feathers','eggs','milk']].any(axis=1).sum())

print("Total numbers of animals that have one of these trait :",total)
```

```python
# 3 Which animal types have the largest number of records


print("Animal types with largest number of records :"
,df['type'].value_counts().idxmax())
print("Number of animals :",df['type'].value_counts().max())
```

```python
# 4  List animals that are both aquatic and predator.
animals = df[df[['aquatic','predator']].all(axis=1)]

print(animals["animal_name"].tolist())
```

```python
# 5 Find the most common characteristics in the dataset
counts = {
    "hair": df["hair"].sum(),
    "feathers": df["feathers"].sum(),
    "eggs": df["eggs"].sum(),
    "milk": df["milk"].sum(),
    "airborne": df["airborne"].sum(),
    "aquatic": df["aquatic"].sum(),
    "predator": df["predator"].sum(),
    "toothed": df["toothed"].sum(),
    "backbone": df["backbone"].sum(),
    "breathes": df["breathes"].sum(),
    "venomous": df["venomous"].sum(),
    "fins": df["fins"].sum(),
    "legs": (df["legs"] > 0).sum(),
    "tail": df["tail"].sum(),
    "domestic": df["domestic"].sum(),
    "catsize": df["catsize"].sum()
}

# Find the most common characteristic
most_common = max(counts, key=counts.get)

print("Most common characteristic:", most_common)
print("Number of animals:", counts[most_common])
```

```python
# 6 Check for missing values and duplicates
ms=df.isnull().sum()
print("Missing values per column:")
print(ms[ms > 0] if ms.sum() > 0 else "No missing values found.")
```

```python
# 7 Use Pandas value_counts and filtering, and NumPy for suitable numeric counts/statistics.

print("--- Last 5 rows of the dataset ---")
display(df.tail())

# Display shape
print(f"\nDataset shape (rows, columns): {df.shape}")

# Display info
print("\n--- Dataset Info ---")
df.info()

# Display statistical description
print("\n--- Statistical Description ---")
display(df.describe())
```

```python
# 8 NumPy and Advanced Pandas Calculations
# (Groupby, Sorting, and NumPy Arrays)

import numpy as np


# Groupby and Sort: Average number of legs per animal type
legs_by_type = df.groupby('type')['legs'].mean().sort_values(ascending=False)
print("Average number of legs per animal type (Sorted):\n", legs_by_type)

legs_array = np.array(legs_by_type.values)
median_legs = np.median(legs_array)
max_legs = np.max(legs_array)

print(f"\nMedian of the average legs across types: {median_legs}")
print(f"Maximum average legs in a type: {max_legs:.2f}")
```

```python
# 9 a bar chart of animal types and a chart comparing selected characteristics

import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style for the plots
sns.set_theme(style="whitegrid")

# Initialize a figure with two subplots side-by-side
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Chart 1: Bar chart of animal types
type_counts = df['type'].value_counts().sort_index()
sns.barplot(x=type_counts.index, y=type_counts.values, ax=axes[0], palette='viridis', hue=type_counts.index, legend=False)
axes[0].set_title('Distribution of Animal Types')
axes[0].set_xlabel('Animal Type')
axes[0].set_ylabel('Count')

# Chart 2: Comparing selected characteristics
features = ['hair', 'feathers', 'eggs', 'milk', 'aquatic', 'predator']
feature_sums = df[features].sum().sort_values(ascending=False)
sns.barplot(x=feature_sums.values, y=feature_sums.index, ax=axes[1], palette='magma', hue=feature_sums.index, legend=False)
axes[1].set_title('Comparison of Selected Animal Characteristics')
axes[1].set_xlabel('Count of Animals with Feature')
axes[1].set_ylabel('Characteristics')

plt.tight_layout()
plt.show()
```
